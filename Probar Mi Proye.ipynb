```python
import pandas as pd
import numpy as np
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from scipy import stats

plt.style.use('ggplot')
sns.set(font_scale=1.1)
```

```python
archivo = "Datos/Diplomado_PARA_PUBLICAR.xlsx"
```

```python
excel = pd.ExcelFile(archivo)
```

```python
print("Hojas disponibles:")
print(excel.sheet_names)
```

```python
if "2022 al 2025 Juntas" in excel.sheet_names:
    df = pd.read_excel(archivo, sheet_name="2022 al 2025 Juntas")
else:
    df = pd.read_excel(archivo, sheet_name=excel.sheet_names[0])
```

```python
print("\nValores faltantes")

faltantes = pd.DataFrame({
    "Variable": df.columns,
    "Faltantes": df.isnull().sum(),
    "Porcentaje": round(df.isnull().mean()*100,2)
})

display(faltantes)
```

```python
print("\nRegistros duplicados")

duplicados = df.duplicated().sum()

print(duplicados)
```

```python
numericas = df.select_dtypes(include=np.number)

print("\nVariables numéricas")

print(numericas.columns.tolist())
```

```python
categoricas = df.select_dtypes(include='object')

print("\nVariables categóricas")

print(categoricas.columns.tolist())
```
